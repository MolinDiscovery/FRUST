```python
import frust as ft

NAME = "test"

method = ft.workflows.methods.preset("wb97xd3-631g")

wf = ft.workflows.screen_ts(
    csv_path="/Users/jacobmolinnielsen/Developer/FrustActivationProject/FRUST/datasets/font_smiles_cat1.csv",
    ts_types=["TS2"],
    method=method,
    n_confs=20,
    top_n=20,
    dft=False,
    ts_backend="tsguess2",
    prune_initial={
        "modes": ("moi", "rmsd"),
        "moi_max_deviation": 0.01,
        "rmsd_max_rmsd": 1.25,
        "rmsd_max_dev": None,
    },
)

stages = [
    stage
    for stage in wf._stage_defs()
    if stage.id not in {"xtb_opt", "dft_pre_sp", "filter", "xtb_sp", "xtb_preopt"}
]

wf._stage_defs = lambda: stages

print(wf.show_stages().to_markdown())
```

```python
df = wf.run(n_cores=10)
```

```python
v = ft.plot_conformers(df, background_color="white", cluster_threshold=0, top_n=20, export_HTML="c.html", columns=2)
```
